**Executive Summary:** Strategic Pricing and Operational Analysis for Big Mountain Resort  
  
**Date:** May 6, 2026  
  
**Subject:** Optimization of Ticket Pricing and Capital Investment Strategy

**1. Problem Statement**

Big Mountain Resort is a ski resort in Montana with 11 lifts, two t-bars, and one magic carpet that is visited by about 350,000 people per year. A recently installed chair lift has added $1,540,000 to operating costs this season.   
  
Currently, the resort charges a premium above average market pricing. However, Big Mountain may not be capitalizing as much as possible with this approach and doesn’t have a good sense of the quality of its facilities compared to others. The Resort needs to increase revenue.  
  
Despite its superior infrastructure and market-leading facilities, the resort has maintained a flat pricing strategy of $81.00. The objective of this analysis was to build a data-driven model to:  

- Determine a market-supported ticket price based on Big Mountain's unique facility profile.

- Quantify the financial impact of various operational changes, such as closing runs or expanding vertical drop.

- Transition from intuition-based pricing to a predictive, competitive model.

**2. Data Wrangling & Exploratory Insights**  

Our analysis began with a comprehensive dataset of 330 resorts nationwide. After rigorous cleaning — including the removal of entries missing target price data and correcting unit errors — we narrowed the scope to 277 resorts for a robust comparison.

**Key Finding:** Big Mountain Resort is a top-tier facility. It ranks in the top percentiles for Vertical Drop, Snow Making Acreage, and Total Chair Count.

**3. The Predictive Model**  
  
We developed a Random Forest Regressor to predict supportable ticket prices. This model outperformed baseline and linear models, capturing complex relationships between a resort's physical assets and its market value.  

- **Accuracy:** The model maintains a Mean Absolute Error (MAE) of approximately $9.00, providing a reliable baseline for pricing decisions.

- **Feature Importance:** The model identified the core "value drivers" for guests.

```python
features = ['Vertical Drop', 'Snow Making Area', 'Total Chairs', 'Fast Quads', 'Runs', 'Longest Run']
importances = [0.45, 0.18, 0.12, 0.08, 0.05, 0.04] # Approximate values from rf_feat_imps

plt.figure(figsize=(10, 6))
plt.barh(features, importances, color='skyblue')
plt.xlabel('Impact on Ticket Price (Importance Score)')
plt.title('What Drives Ticket Value? (Top Features)')
plt.gca().invert_yaxis()
plt.show()
```

**4. Market Positioning & Results**  

The model predicts that Big Mountain Resort's facilities support a ticket price of $95.87. 

This indicates that the resort is currently underpriced by approximately 18.4% ($14.87 per ticket).

**Scenario Analysis:** Impact of Operational Decisions  
We simulated "What-If" scenarios to understand the financial consequences of management choices.  

**Scenario 1: Closing Runs**  
Closing runs reduces the perceived value of the resort. While closing 1 run has a negligible impact, closing 5 or more runs leads to a sharp decline in supportable revenue.  

**Scenario Modeling: Revenue Impact of Closing Runs**

```python
runs_closed = [0, 1, 2, 3, 4, 5]
revenue_change = [0, 0, -525000, -1250000, -2100000, -3200000]

plt.figure(figsize=(10, 5))
plt.plot(runs_closed, revenue_change, marker='o', linestyle='-', color='red')
plt.xlabel('Number of Runs Closed')
plt.ylabel('Change in Seasonal Revenue ($)')
plt.title('Risk Analysis: Revenue Impact of Run Closures')
plt.grid(True, alpha=0.3)
plt.show()
```

**5. Strategic Recommendations**  

**1. Immediate Pricing Adjustment:** Align the ticket price closer to the market-supported value. Increasing the price from `$81.00`
to `$94.00` remains conservative relative to the model's $95.87 prediction while significantly boosting revenue.

**Maximized Capacity:** Maintain all 105 runs throughout the peak season. The data shows that the "Run Count" is a primary psychological and physical driver of price support.

**Season Extension:** Every additional day of operation is estimated to generate $1.73M in incremental revenue. There's no need to invest in additional snowmaking acreage as Big Mountain Resort is already a leader in snowmaking, so increased snowmaking leads to negligible potential profit. Prioritize staff retention to extend the season by even 2–3 days.

**6. Conclusion & Future Scope**  

Big Mountain Resort is currently a market leader charging "middle-of-the-pack" prices. By leveraging its infrastructure — specifically its vertical drop and capacity — the resort can justify a premium price point.
